### Dropping unnecessary columns

```python
df_1 = df.drop(columns=["Unnamed: 0", "id", "track_name", "ver"])
df_1
```

### Removing apps with zero app rating

```python
df_1 = df_1[df_1["user_rating"] > 0]
df_1
```

### Creating Helping columns

```python
# Free vs Paid
df_1["is_paid"] = df_1["price"].apply(lambda x: 1 if x > 0 else 0)

# Log transform skewed popularity
df_1["log_rating_count"] = np.log1p(df_1["rating_count_tot"])
df_1
```

```python
## Plotting for visualisation
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df_1["rating_count_tot"], bins=50)
plt.title("Original Rating Count Distribution")
plt.show()

sns.histplot(df_1["log_rating_count"], bins=50)
plt.title("Log Transformed Rating Count Distribution")
plt.show()
```
